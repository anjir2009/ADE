```python
dbutils.fs.mkdirs("/Volumes/workspace/data_analytics/sales_volume/raw")
```

```python
dbutils.fs.put("/Volumes/workspace/default/new_volume/new_folder/hello.txt", "Hello from DBFS!", overwrite=True)
```
